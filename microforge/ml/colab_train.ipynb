## 1. Récupérer le dataset
Téléverser `dataset.zip` (export de `data/dataset`, format YOLO) ou le tirer depuis le remote DVC.

```python
from google.colab import files

up = files.upload()  # dataset.zip
!unzip -q dataset.zip -d dataset
```

```python
import pathlib

import yaml

cfg = yaml.safe_load(open("dataset/dataset.yaml"))
cfg["path"] = str(pathlib.Path("dataset").resolve())
yaml.safe_dump(cfg, open("dataset/dataset.yaml", "w"))
print(len(cfg["names"]), "classes")
```

## 2. Entraîner

```python
from ultralytics import YOLO

model = YOLO("yolo11s.pt")
res = model.train(data="dataset/dataset.yaml", epochs=80, imgsz=640, batch=16)
print(res.results_dict)
```
